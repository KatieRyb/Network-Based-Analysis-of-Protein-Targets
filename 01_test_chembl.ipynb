```python
#basic test that code runs
#print("Hello computational biology!")

import requests
import pandas as pd
drug = "ibuprofen"

url = f"https://www.ebi.ac.uk/chembl/api/data/molecule/search?q={drug}&format=json"

response = requests.get(url)

data = response.json()

data

data.keys()
#output up to here: 
#dict_keys(['molecules', 'page_meta'])

data["molecules"][0]

chembl_id = data["molecules"][0]["molecule_chembl_id"]

chembl_id
```

```python
target_url = (
    f"https://www.ebi.ac.uk/chembl/api/data/mechanism"
    f"?molecule_chembl_id={chembl_id}&format=json"
)

target_response = requests.get(target_url)

target_data = target_response.json()

target_data
```

```python
target_id = target_data["mechanisms"][0]["target_chembl_id"]

target_id

protein_url = f"https://www.ebi.ac.uk/chembl/api/data/target/{target_id}.json"

protein_response = requests.get(protein_url)

protein_data = protein_response.json()

protein_data
```

```python
components = protein_data["target_components"]

for protein in components:
    print(
        protein["component_description"],
        protein["accession"]
    )
```

```python
import pandas as pd

rows = []

for protein in components:
    rows.append({
        "Drug": "Ibuprofen",
        "Protein": protein["component_description"],
        "UniProt": protein["accession"]
    })

df = pd.DataFrame(rows)

df
```
