```python
import pandas as pd
import numpy as np
```

```python
#create binary drug-pathway matrix

#drug x pathway

"""

Drug → protein target → Reactome pathway


 use this matrix to calculate molecular/pathway similarity.

"""

file_path_drug_pathways = "C:/Katieryb/Pipelines/proteins/drug_pathways.csv"

drug_pathways = pd.read_csv(file_path_drug_pathways)

drug_pathway_matrix = pd.crosstab(
    drug_pathways["Drug"],
    drug_pathways["Pathway_ID"]
)

# Convert counts to binary presence/absence
drug_pathway_matrix = (drug_pathway_matrix > 0).astype(int)

print(drug_pathway_matrix.shape)

drug_pathway_matrix.head()
```
